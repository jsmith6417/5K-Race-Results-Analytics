# **Combine Race Results**

Combines cleaned 5K@ADA and 5K@EASD race results for visualization.

```python
# Import packages

# For data manipulation
import pandas as pd

# For working with files
import glob
import os

# For working with system commands
import sys
```

```python
# Define custom NA values, excluding 'NA' to make sure that Country Code NA (Namibia) is not interpreted as missing data
custom_na_values = ['N/A', 'NaN', 'null', '']
```

```python
# Find the latest ADA data file

# Define the directory and filename pattern
directory = r"D:\OneDrive\Documents\Python\Current\Race Results\ADA"
pattern = "ADA Race Data for Visualization.csv"

# Construct the full search pattern
search_pattern = os.path.join(directory, pattern)

# Find all matching files
matching_files = glob.glob(search_pattern)

# Ensure there is at least one match
if not matching_files:
    raise SystemExit("No matching files found. Exiting program.")

# Sort files by modification time to get the most recent one
latest_file = max(matching_files, key=os.path.getmtime)

# Extract just the file name
file_name = os.path.basename(latest_file)

print(f"Found file: {file_name}")
```

```python
# Load ADA data
df0 = pd.read_csv(latest_file, keep_default_na=False, na_values=custom_na_values)
```

```python
# Display the first 5 rows of df0
df0.head()
```

```python
# Display basic information about df0
df0.info()
```

```python
# Display the number of missing values in each column
df0.isnull().sum()
```

```python
# Find the latest EASD data file

# Define the directory and filename pattern
directory = r"D:\OneDrive\Documents\Python\Current\Race Results\EASD"
pattern = "EASD Race Data for Visualization.csv"

# Construct the full search pattern
search_pattern = os.path.join(directory, pattern)

# Find all matching files
matching_files = glob.glob(search_pattern)

# Ensure there is at least one match
if not matching_files:
    raise SystemExit("No matching files found. Exiting program.")

# Sort files by modification time to get the most recent one
latest_file = max(matching_files, key=os.path.getmtime)

# Extract just the file name
file_name = os.path.basename(latest_file)

print(f"Found file: {file_name}")
```

```python
# Load EASD data
df1 = pd.read_csv(latest_file, keep_default_na=False, na_values=custom_na_values)
```

```python
# Display the first 5 rows of df1
df1.head()
```

```python
# Display basic information about df1
df1.info()
```

```python
# Display the number of missing values in each column
df1.isnull().sum()
```

```python
# Combine df0 and df1 into a single DataFrame
df2 = pd.concat([df0, df1], ignore_index=True)
```

```python
# Convert HH:MM:SS to a numeric fraction of a day
df2['Time'] = pd.to_timedelta(df2['Time']).dt.total_seconds() / 86400
```

```python
# Display the first 5 rows of df2
df2.head()
```

```python
# Display basic information about df2
df2.info()
```

```python
# Display the number of missing values in each column
df2.isnull().sum()
```

```python
# Save the cleaned data to a new csv file
df2.to_excel('Race Data for Visualization.xlsx', sheet_name='Results', index=False)
```
